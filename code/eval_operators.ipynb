```python
# the box plots:
import plotly.express as px

def plot_box(im_name, im_df, cross, mut):
    # filter DFs:
    print(f"creating boxplot_impact_{im_name}_{cross}_{mut}.pdf")
    filtered_df = im_df[(im_df['crossover'] == cross) & (im_df['mutation'] == mut)].drop(columns=['generation', 'run_number', 'crossover', 'mutation'], axis=1)

    filtered_df.columns = [str(col) if col != 'initial_population' else col for col in filtered_df.columns]



    fig = px.box(filtered_df, facet_col='initial_population', facet_col_spacing=0.01 )
    fig.update_traces(
        line_width=.75,
        marker={'size': 3},
        width=0.6
    ) 

    #set axis titles:
    fig.update_yaxes(title_text=im_name, row=1, col=1)
    fig.update_xaxes(title_text='traceID')


    #for some dumb reason I have to set the layout manually here
    fig.update_layout(
        font={'size': footnotesize, 'family': 'Libertinus', 'color': 'black'},
        title={'font': {'size': normalsize}},
        plot_bgcolor=plotly_default.layout.plot_bgcolor,
        paper_bgcolor=plotly_default.layout.paper_bgcolor,
        legend={
            'font': {'size': footnotesize},
            'title': {'font': {'size': footnotesize}},
        },
        width=dis_wide_figure_size,
        height=150,
        margin=dict(l=30, r=10, t=20, b=50),
        autosize=False
    )

    fig.update_xaxes(
        title={'font': {'size': normalsize}, 'standoff': 5},
        tickfont={'size': 5},
        gridcolor=grid_color,
        linecolor=bordercolor,
        gridwidth=grid_size,
        linewidth=border_size,
        mirror=True,
        zerolinecolor=zeroline_color,
        zerolinewidth=zeroline_size,
        showline=True,
        tickmode='linear', 
        tick0=0, 
        dtick=1
    )

    fig.update_yaxes(
        title={'font': {'size': normalsize}, 'standoff': 5},
        tickfont={'size': footnotesize},
        gridcolor=grid_color,
        linecolor=bordercolor,
        gridwidth=grid_size,
        linewidth=border_size,
        mirror=True,
        zerolinecolor=zeroline_color,
        zerolinewidth=zeroline_size,
        showline=True
    )

    fig.update_xaxes(tickmode='linear', tick0=0, dtick=1)  # Zeigt jeden Tick

    fig.for_each_annotation(lambda a: a.update( text=a.text.replace("initial_population=", "") ))
    name = im_name.replace(" ", "_")
    
    if im_name == "$I_C$":
        name="counting_impact"
    elif im_name == "$I_F$":
        name="fitness_impact"
    elif im_name == "$I_E$":
        name="entropy_impact"
    elif im_name == "$I_{F \\cdot E}$":
        name="fitness_entropy_impact"
    fig.write_image(f"../figures/single_run_in_detail/boxplot_impact_{name}_{cross}_{mut}.pdf")




populations = ci_df['initial_population'].unique()
crossovers = ci_df['crossover'].unique()
mutations = ci_df['mutation'].unique()

impact_metrics = {
    "$I_C$": ci_df,
    "$I_F$": fi_df,
    "$I_E$": ei_df,
    "$I_{F \\cdot E}$": fei_df
}

for im_name, im_df in impact_metrics.items():
    for co in crossovers:
        for mu in mutations:
            plot_box(im_name, im_df, co, mu)
```

# Table comparing Performance Indicators

```python
import pandas as pd

counting_impact = pd.read_csv("../data/test_results/counting_impact_pop.csv.zip") #NOTE: Change back to .csv if you generated the data yourself
fitness_impact = pd.read_csv("../data/test_results/fitness_impact_pop.csv.zip")
entropy_impact = pd.read_csv("../data/test_results/entropy_impact_pop.csv.zip")
fitness_entropy_impact = pd.read_csv("../data/test_results/fitness_entropy_impact_pop.csv.zip")

gen = 75

counting_impact = counting_impact[counting_impact['generation'] == gen]
fitness_impact = fitness_impact[fitness_impact['generation'] == gen]
entropy_impact = entropy_impact[entropy_impact['generation'] == gen]
fitness_entropy_impact = fitness_entropy_impact[fitness_entropy_impact['generation'] == gen]
```

```python
from scipy.stats import mannwhitneyu

impact_data = {
    "CI" : counting_impact,
    "FI" : fitness_impact,
    "EI" : entropy_impact,
    "FEI" : fitness_entropy_impact
}


decimal_places = 3
#This table was very vibe-coded with:
#build one table for each of the initial_populations! The rows should be the traceIDs, and the columns should be the average values for the four impact metrics. Also, calculate significant differences, and mark significant different resutls with a * in the table:
rows = []
for initial_population in counting_impact['initial_population'].unique():
    for trace_id in [col for col in counting_impact.columns if 'traceID_' in col]:
        row = {"traceID": trace_id.replace('traceID_', '')}
        for impact_name, df in impact_data.items():
            median_value = df[df['initial_population'] == initial_population][trace_id].median()
            row[f"{impact_name}"] = median_value

            #calculate significant difference to CI
            if impact_name != "CI":
                ci_values = counting_impact[counting_impact['initial_population'] == initial_population][trace_id]
                other_values = df[df['initial_population'] == initial_population][trace_id]
                stat, p = mannwhitneyu(ci_values, other_values)
                if p < 0.05:
                    row[f"{impact_name}"] = f"{round(median_value,decimal_places)}*"
        rows.append(row)
    filtered_df = pd.DataFrame(rows)
    print(f"Initial Population: {initial_population}")
    
    #switch rows and cols:
    filtered_df = filtered_df.set_index('traceID')  # traceID als Index setzen
    filtered_df = filtered_df.T  # Transponieren
    filtered_df.to_latex(f"../figures/impact_summary_initial_pop_{initial_population}.tex", float_format=f"%.{decimal_places}f")
    rows = [] #reset rows
```
